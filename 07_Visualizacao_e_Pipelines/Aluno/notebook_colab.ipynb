## 1. Matplotlib: Gráficos Básicos

Todo gráfico em Matplotlib começa com a mesma receita: `fig, ax = plt.subplots()`, depois `ax.plot(...)`/`ax.bar(...)`/`ax.scatter(...)`, e por fim `plt.show()`. `fig` é a **figura inteira** (a "folha", que você exporta/salva); `ax` são os **eixos** — a área de desenho dentro da figura, onde o gráfico realmente aparece. No dia a dia, você desenha em cima do `ax`, e só usa o `fig` quando precisa mexer na figura inteira (como exportar, Seção 2).

> 💡 Por que `fig, ax` desde o início? Existe um jeito mais curto (`plt.plot(...)` direto), mas só funciona bem para 1 gráfico isolado. O padrão `fig, ax` é o que você vai usar sempre — inclusive com vários gráficos juntos (Seção 2).

**Qual gráfico usar quando:**

| Tipo de gráfico | Quando usar |
|---|---|
| Linha (`ax.plot`) | Evolução ao longo do tempo |
| Barra (`ax.bar`) | Comparar categorias |
| Dispersão (`ax.scatter`) | Relação entre 2 variáveis numéricas |

### 🔹 Exemplo 1 — Gráfico de linha

```python
import matplotlib.pyplot as plt
import pandas as pd

vendas = pd.DataFrame({
    "mes": ["Jan", "Fev", "Mar", "Abr", "Mai"],
    "receita": [1200, 1500, 900, 1800, 2100],
})

fig, ax = plt.subplots()
ax.plot(vendas["mes"], vendas["receita"])
plt.show()
```

### 🔹 Exemplo 2 — Gráfico de barras

```python
vendas_categoria = pd.DataFrame({
    "categoria": ["Papelaria", "Acessórios", "Eletrônicos"],
    "receita_total": [3200, 1800, 4500],
})

fig, ax = plt.subplots()
ax.bar(vendas_categoria["categoria"], vendas_categoria["receita_total"])
plt.show()
```

### 🔹 Exemplo 3 — Gráfico de dispersão

```python
import matplotlib.pyplot as plt
import pandas as pd

vendas = pd.DataFrame({
    "mes": ["Jan", "Fev", "Mar", "Abr", "Mai"],
    "receita": [1200, 1500, 900, 1800, 2100],
})

fig, ax = plt.subplots()
ax.scatter(vendas["receita"], [10, 14, 8, 18, 21])  # ex.: receita x nº de pedidos
plt.show()
```

## 2. Customização, Subplots e Exportação

### 🔹 Exemplo 1 — Customizando um gráfico

📖 **Antes do código:** `color` aceita um código hexadecimal (`#RRGGBB` — 2 dígitos para vermelho, 2 para verde, 2 para azul, de `00` a `ff`; `#CC0000` é um vermelho forte) ou nomes simples como `"red"`/`"blue"`. `marker` desenha um símbolo em cada ponto do gráfico — `"o"` é um círculo; outras opções comuns são `"s"` (quadrado) e `"^"` (triângulo). Nenhum dos dois é obrigatório — sem eles, o Matplotlib usa cor e estilo padrão.

```python
import matplotlib.pyplot as plt
import pandas as pd

vendas = pd.DataFrame({
    "mes": ["Jan", "Fev", "Mar", "Abr", "Mai"],
    "receita": [1200, 1500, 900, 1800, 2100],
})

fig, ax = plt.subplots()
ax.plot(vendas["mes"], vendas["receita"], color="#CC0000", marker="o", label="Receita")
ax.set_title("Receita mensal")
ax.set_xlabel("Mês")
ax.set_ylabel("Receita (R$)")
ax.legend()
plt.show()
```

### 🔹 Exemplo 2 — Subplots: vários gráficos lado a lado

📖 **Antes do código:** até aqui, `plt.subplots()` sempre devolveu 1 `ax` só (Seção 1). Quando você pede mais de 1 gráfico — `plt.subplots(1, 2)`, por exemplo — ele devolve uma **lista de eixos**, guardada em `axs` (com "s", de plural). Por isso você acessa cada gráfico individual com `axs[0]`, `axs[1]`, etc. — a mesma ideia de indexar uma lista que você já usa desde a Semana 02.

```python
import matplotlib.pyplot as plt
import pandas as pd

vendas = pd.DataFrame({
    "mes": ["Jan", "Fev", "Mar", "Abr", "Mai"],
    "receita": [1200, 1500, 900, 1800, 2100],
})
vendas_categoria = pd.DataFrame({
    "categoria": ["Papelaria", "Acessórios", "Eletrônicos"],
    "receita_total": [3200, 1800, 4500],
})

fig, axs = plt.subplots(1, 2, figsize=(10, 4))  # 1 linha, 2 colunas de gráficos

axs[0].plot(vendas["mes"], vendas["receita"])
axs[0].set_title("Evolução mensal")

axs[1].bar(vendas_categoria["categoria"], vendas_categoria["receita_total"])
axs[1].set_title("Por categoria")

plt.show()
```

### Exportação de gráficos

`fig.savefig()` salva a figura inteira (não a tela) em um arquivo de imagem — essencial para colocar o gráfico num relatório ou apresentação.

```python
import matplotlib.pyplot as plt
import pandas as pd

vendas = pd.DataFrame({
    "mes": ["Jan", "Fev", "Mar", "Abr", "Mai"],
    "receita": [1200, 1500, 900, 1800, 2100],
})

fig, ax = plt.subplots()
ax.plot(vendas["mes"], vendas["receita"])
fig.savefig("receita_mensal.png", dpi=150)
print("Gráfico exportado.")
```

## 3. Funções Reutilizáveis e Introdução a Classes

Você já cria funções desde a Semana 04. Mas há situações em que uma função não basta: quando você precisa agrupar **dados** e **comportamento** juntos.

### 🔹 Exemplo 1 — Função de transformação reutilizável

📖 **Antes do código:** a função abaixo aplica a fórmula de **normalização min-max** — ela pega cada valor de uma coluna e recalcula numa escala de 0 a 1, usando `(valor - mínimo) / (máximo - mínimo)`. O menor valor da coluna vira 0, o maior vira 1, e os demais ficam proporcionalmente entre os dois. É útil quando você quer comparar colunas com escalas muito diferentes (ex.: receita em milhares x nota de 0 a 5) de forma justa.

```python
import pandas as pd

vendas = pd.DataFrame({
    "mes": ["Jan", "Fev", "Mar", "Abr", "Mai"],
    "receita": [1200, 1500, 900, 1800, 2100],
})

def normalizar(coluna):
    return (coluna - coluna.min()) / (coluna.max() - coluna.min())

vendas["receita_normalizada"] = normalizar(vendas["receita"])
print(vendas)
```

### 🔹 Exemplo 2 — Sua primeira classe

📖 **Antes do código:** uma **classe** é um molde — define o que todo objeto desse tipo vai ter (dados) e o que ele vai saber fazer (comportamento). Um **objeto** é uma "cópia preenchida" desse molde, com valores próprios.

| Termo | O que é |
|---|---|
| `class Produto:` | Define o "molde" — o que todo Produto vai ter/fazer |
| `__init__` | Roda automaticamente ao criar um Produto novo, para preencher os dados iniciais |
| `self` | Representa "este Produto específico" — é assim que cada objeto guarda seus próprios dados |
| `mochila = Produto(...)` | Cria um **objeto** — uma cópia preenchida do molde |
| **Método** | O nome que se dá para uma função definida dentro de uma classe (ex.: `aplicar_desconto`) |

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco

    def aplicar_desconto(self, percentual):
        self.preco = self.preco * (1 - percentual / 100)

mochila = Produto("Mochila", 150.0)
mochila.aplicar_desconto(10)
print(mochila.nome, mochila.preco)
```

💡 **Função x Classe:** uma função recebe dados, processa, devolve um resultado, e esquece tudo depois. Uma classe **guarda** dados (no `self`) e empacota junto as ações que fazem sentido para aqueles dados.

> ⚠️ **Veja como é um erro real do Python.** Se você esquecer o `self` ao definir um método, ao chamar `Produto("Caderno", 12.5).mostrar()` você vai ver `TypeError: Produto.mostrar() takes 0 positional arguments but 1 was given`.

```python
class ProdutoComBug:
    def mostrar():  # faltou o self
        print("produto")

try:
    ProdutoComBug().mostrar()
except TypeError as e:
    print(f"Erro: {e}")
```

## 4. ETL e Integração Python-SQL

**ETL** significa **E**xtract (extrair) → **T**ransform (transformar) → **L**oad (carregar). Você já faz isso desde a Semana 04 — só não tinha esse nome:

| Etapa | O que você já aprendeu que se encaixa aqui |
|---|---|
| **E**xtract | Ler um arquivo (`pd.read_csv`) |
| **T**ransform | Limpar e transformar com Pandas (Semana 06) |
| **L**oad | Salvar o resultado — em arquivo ou em banco de dados |

💡 **E as classes que você acabou de aprender?** Este pipeline é pequeno, então uma função dá conta do recado — é mais direto. Mas quando um pipeline cresce (várias fontes de dados, várias transformações, configuração para lembrar), empacotar tudo numa classe (como você fez com `Produto`) ajuda a organizar — é a mesma ideia, só aplicada em maior escala. Guarde essa conexão para quando os pipelines ficarem mais complexos.

### 🔹 Exemplo 1 — Um pipeline ETL completo

```python
import pandas as pd

vendas = pd.DataFrame({
    "mes": ["Jan", "Fev", "Mar", "Abr", "Mai"],
    "receita": [1200, 1500, 900, 1800, 2100],
})

def normalizar(coluna):
    return (coluna - coluna.min()) / (coluna.max() - coluna.min())

vendas.to_csv("vendas_bruto.csv", index=False)               # (preparando o exemplo)

dados = pd.read_csv("vendas_bruto.csv")                       # E - Extract
dados["receita_normalizada"] = normalizar(dados["receita"])  # T - Transform
dados.to_csv("vendas_tratado.csv", index=False)               # L - Load (em arquivo)
print("Pipeline concluído.")
```

### 🔹 Exemplo 2 — Load num banco de dados com sqlite3

📖 **Antes do código:** depois de carregar os dados numa tabela, você pode consultá-los com **SQL** — a linguagem usada para conversar com bancos de dados. Neste exemplo, `SELECT * FROM vendas WHERE receita > 1000` lê quase como uma frase: "selecione tudo (`*`) da tabela `vendas` onde a receita é maior que 1000". Você vai aprender SQL de verdade, a fundo, no Bloco 5 do curso (Semanas 09-10) — aqui é só uma prévia de como Python e SQL conversam.

```python
import pandas as pd
import sqlite3

vendas = pd.DataFrame({
    "mes": ["Jan", "Fev", "Mar", "Abr", "Mai"],
    "receita": [1200, 1500, 900, 1800, 2100],
})

def normalizar(coluna):
    return (coluna - coluna.min()) / (coluna.max() - coluna.min())

vendas.to_csv("vendas_bruto.csv", index=False)
dados = pd.read_csv("vendas_bruto.csv")                       # repetindo o Extract do Exemplo 1
dados["receita_normalizada"] = normalizar(dados["receita"])  # repetindo o Transform do Exemplo 1

conexao = sqlite3.connect("dados.db")
dados.to_sql("vendas", conexao, index=False, if_exists="replace")  # L - Load (em banco)

resultado = pd.read_sql("SELECT * FROM vendas WHERE receita > 1000", conexao)
print(resultado)

conexao.close()
```

💡 **Por que `sqlite3` e não PostgreSQL aqui?** `sqlite3` não precisa de servidor nem configuração. No mercado, o mesmo `pandas.to_sql()`/`pd.read_sql()` funciona com PostgreSQL (via `SQLAlchemy`), só muda a conexão.
